# 📓 01: Data Exploration and Validation Notebook

This notebook contains the exploratory data analysis (EDA) and data quality validation checks for the **Social Media Sentiment Analysis** dataset (`data/raw_tweets.csv`).

## Setup and Libraries

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
import os

# Add source directory to path
sys.path.append(os.path.abspath("../src"))
from data_validator import DataValidator
```

## 1. Load Data

```
csv_path = "../data/raw_tweets.csv"
df = pd.read_csv(csv_path)
df.head()
```

## 2. Profile Quality & Schema Verification

```
# Validate schema
valid, msg = DataValidator.validate_csv_schema(csv_path, ["text", "sentiment"])
print(f"Schema Valid: {valid} ({msg})")

# Profile quality metrics
profile = DataValidator.profile_dataset(df, "text", "sentiment")
for k, v in profile.items():
    if k != "label_distribution":
        print(f"{k}: {v}")
```

## 3. Visualize Sentiment Distribution

```
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='sentiment', palette='viridis')
plt.title('Distribution of Sentiment Labels')
plt.xlabel('Sentiment Class')
plt.ylabel('Count')
plt.show()
```

## 4. Text Character and Word Length Analysis

```
df['char_length'] = df['text'].astype(str).str.len()
df['word_count'] = df['text'].astype(str).str.split().apply(len)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.histplot(data=df, x='char_length', hue='sentiment', multiple='stack', ax=axes[0], bins=30, kde=True)
axes[0].set_title('Character Length Distribution')

sns.histplot(data=df, x='word_count', hue='sentiment', multiple='stack', ax=axes[1], bins=30, kde=True)
axes[1].set_title('Word Count Distribution')
plt.show()
```
